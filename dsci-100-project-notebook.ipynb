#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.

```r
summary_stats <- basketball_tidy |>
     summarise(across(c(seed, reg_wins_pct, conf_wins_pct),
                   list(mean = ~mean(.x, na.rm = TRUE),
                        max  = ~max(.x, na.rm = TRUE),
                        min  = ~min(.x, na.rm = TRUE))))
summary_stats
```

#### Task 4.4: Plot the distribution of two relevant variables in the raw data (4 points)

```r
numeric_plot <- basketball_tidy |>
    ggplot(aes(x = reg_wins_pct)) +
    geom_histogram(binwidth = 5) +
    xlab("Regional % Wins") +
    ylab("Number of Teams") +
    labs(title = "Distribution of Percentage of Regional Wins") +
    theme(text = element_text(size = 15))
numeric_plot
```

```r
basketball_tidy_full <- basketball_data |>
    select(seed, reg_wins_pct, conf_wins_pct, tourney_finish)
tourney_finish_counts <- basketball_tidy_full |>
    count(tourney_finish)
categorical_plot <- tourney_finish_counts |>
    ggplot(aes(x = tourney_finish, y = n)) +
    geom_bar(stat = "identity") +
    xlab("Tourney Finish Place") +
    ylab("Number of Teams") +
    labs(title = "Distribution of Tourney Finishing Places") +
    theme(text = element_text(size = 15))
categorical_plot
```

Visualization 1: This visualization shows the relationship between the seed of a team and their the percentage of conference wins they had that year.
